```python
from PIL import Image
import os

# ASCII characters used to represent pixels
ASCII_CHARS = "@$B%8&WM#*oahkbdpqwmZO0QLCJUYXzcvunxrjft/\\|()1{}[]?-_+~<>i!lI;:,\"^`'. "

def resize(image, new_width=80):
    width, height = image.size
    ratio = height / width / 1.65  # adjust aspect ratio
    new_height = int(new_width * ratio)
    return image.resize((new_width, new_height))

def grayify(image):
    return image.convert("L")

def pixels_to_ascii(image):
    pixels = image.getdata()
    chars = "".join([ASCII_CHARS[min(pixel // 25, len(ASCII_CHARS) - 1)] for pixel in pixels])
    return chars

def image_to_ascii(image_path, width=80):
    image = Image.open(image_path)
    image = resize(image, width)
    image = grayify(image)
    ascii_str = pixels_to_ascii(image)
    img_width = image.width
    ascii_str = "\n".join([ascii_str[i:i+img_width] for i in range(0, len(ascii_str), img_width)])
    return ascii_str

# Convert all frames
output = []
frames_dir = "frames"
for file in sorted(os.listdir(frames_dir)):
    if file.endswith(".png"):
        ascii_frame = image_to_ascii(os.path.join(frames_dir, file), width=80)
        output.append(ascii_frame)

# Save all frames to JSON (for your website)
import json
with open("ascii_frames.json", "w") as f:
    json.dump(output, f)
```
